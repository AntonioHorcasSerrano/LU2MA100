La nomenclature adoptée est la classique, pour le nombre de parties de $k$ éléments dans un ensemble à $n$ éléments. La matrice carrée d'ordre $n+1$ se construit par récurrence de la forme:
$$\begin{pmatrix} n \\ k \end{pmatrix}=\begin{pmatrix} n-1 \\ k-1 \end{pmatrix}+\begin{pmatrix} n-1 \\ k \end{pmatrix}\quad (0<k<n)$$ 

```python
def PascalTriangle(n):
    # Set up a n+1 x n+1 null matrix
    # T = [[0] * (n+1) for i in range(n+1)], or:
    T = [[0 for x in range(n+1)] for x in range(n+1)] 
    for n in range(n+1): # for each line
        if n == 0:
            T[n][0] = 1
        else:
            for k in range(n+1): # for each column
                if k == 0:
                    T[n][0] = 1
                else:
                    T[n][k] = T[n-1][k-1] + T[n-1][k]
    return T
T = PascalTriangle(5)
```
