```python
import numpy as np;
import matplotlib.pyplot as plt;
```

```python
def fractal(P, d):
    if d == 0:
        return P
    P = np.linspace(P[0], P[1], 5)
    v = P[3] - P[1]
    P[2] += np.sqrt(3)/2*np.array([-v[1], v[0]])
    S1 = fractal(P[1:3], d-1)
    S2 = fractal(P[2:4], d-1)
    S3 = np.vstack((P[0], S1[:-1], S2, P[-1]))
    return S3
```
