# Exercice 1: Calcul du diamètre d'un graphe

Dans un graphe $G$, la distance entre deux sommets $i$ et $j$ est le nombre minimal d'arêtes séparant ces deux sommets. Le diamètre de $G$ est la distance maximale séparant deux sommets de $G$. Le but de cet exercice est de calculer le diamètre d'un graphe.

Dans la suite, on suppose que tous nos graphes sont connexes (c'est-à-dire qu'il existe un chemin entre toute paire de sommets) et possèdent au moins deux sommets.

On rappelle que si $G=(X,E)$ est un graphe avec $X=\{1,2,\dots,n\}$, sa matrice d'adjacence $A\in\mathcal{M}_{n,n}(\mathbb{R})$ est définie par :

$$
A_{i,j}=\begin{cases}
1 & \text{si } \{i,j\} \in E\\ 
0 & \text{si } \{i,j\} \notin E \,.
\end{cases}
$$

Il est possible de montrer par récurrence sur $k\geq 0$ que $A^k_{i,j}\neq 0$ si et seulement si il existe un chemin de longueur $k$ de $i$ à $j$ dans $G$. En fait $A^k_{i,j}$ est exactement le nombre de chemins de $i$ à $j$ de longueur $k$.

**a)**
Écrire deux fonctions `complet(n)` et `cycle(n)` qui prennent en argument un entier $n$ et renvoient la matrice d'adjacence de respectivement : $K_n$ le graphe complet à $n$ sommets, et $C_n$ le cycle à $n$ sommets.



La matrice d'adjacence du graphe complet est définie: $a_{ij}=1$ si $i\neq j$, $0$ sinon
La matrice d'adjacence du graphe cycle est définie; $a_{ij}=1$ si $|i-j|=1$ ou $|i-j|=n-1$, $0$ sinon

```python
def complet(n):
    A = np.ones((n,n))
    np.fill_diagonal(A,0)
    return A

def cycle(n):
    A = np.zeros((n,n))
    for i in range(n):
        for j in range(n):
            if np.abs(i-j) == 1 or np.abs(i-j) == n-1:
                A[i][j] = 1
    return A    
```


**b)**
Écrire une fonction `positifs_coeffs(M)` qui prend en argument une matrice $M$ et renvoie `True` si tous ses coefficients non diagonaux sont strictement positifs, et `False` sinon. À noter que les coefficients de $M$ sont automatiquement plus grands ou égaux à zéro.



```python
# M matrice carré car représente graphe
def positifs_coeffs(M):
    display(M)
    n = M.shape[0]
    for i in range(n):
        for j in range(n):
            if i!=j:
                if M[i][j] == 0:
                    return False
    # si arrivé ici tous les coefficients non diagonaux sont strictement positifs
    return True
```


**c)**
Écrire une fonction `diametre(A)` qui, étant donné un graphe défini par sa matrice d'adjacence $A$, renvoie son diamètre.

**Indication:**
Considérer pour $k\geq 1$ la matrice $M_k=A+A^2+\dots+A^k$.



```python
def diametre(A):
    M = A.copy()
    k = 1
    while not positifs_coeffs(M):
        k+=1
        M = M + np.linalg.matrix_power(A,k)
    return k
```

# Exercice 2: Exponentielle de matrices

Le but de cet exercice est de développer un algorithme permettant de calculer l'exponentielle d'une matrice réelle carrée. Si $A$ est une matrice réelle carrée, son exponentielle est définie par la série:

$$
\exp(A) = \sum_{k=0}^\infty \frac{A^k}{k!} \,,
$$

par analogie avec l'exponentielle sur les nombres réels. Ici $A^k$ représente le produit matriciel de $A$ avec elle-même $k$ fois.

**a)**
Définir les tableaux Numpy correspondant aux matrices $A_1$ et $A_2$ définies par:

$$

0 & 1.2 & 0.9
\end{pmatrix} ,
&
A_{2}	&= \begin{pmatrix}
2 & 3 & 2\\ 
1 & 2 & 3\\ 
4 & 3 & 5.2
\end{pmatrix} .
\end{align*}
$$

```python
# on définie directement les matrices
A1 = np.array([[1,.8,.6],[.8,.2,.8],[0,1.2,.9]])
A2 = np.array([[2,3,2],[1,2,3],[4,3,5.2]])
print(A1)
print(A2)
```


**b)**
Définir une fonction `matrix_power(A,n=20)` retournant une approximation de $\exp(A)$ obtenue en gardant uniquement les $n+1$ premiers termes de la série, *i.e.* les termes de $k=0$ à $k=n$.



```python
import math

def matrix_power(A, n=20):
    # initialize l'exponentielle aprochée avec la matrice identité de même taille
    expA = np.identity(A.shape[0])
    for k in range(1, n+1):
        expA += np.linalg.matrix_power(A,k)/math.factorial(k)
    return expA
```

$$
\Vert A\Vert_{F}^{2}=\operatorname{tr}(AA^{t})=\sum_{i=1}^{m}\sum_{j=1}^{m}|a_{ij}|^{2} \,.
$$

```python
def norm_frobenius(A):
    m = A.shape[0]
    norme = 0
    for i in range(m):
        for j in range(m):
            norme = norme + np.abs(A[i][j])**2
    return np.sqrt(norme)
```
